```python
import random
import time


# --------------------
# Game State
# --------------------

game_state = {
    "player": 0,
    "dealer": 0,
    "game_over": False
}

# --------------------
# Variables
# --------------------


try_again = "yes"


# --------------------
# Fuctions
# --------------------

def create_cards_deck():

     """ Create a new deck """

     return (list(range(2,11)) + ['J','Q','K','A']) * 4

def score_calculate(cards):

    """Calculate the player score."""
    x = 0
    for card in cards:
        if card in ['J','Q','K']:
            cards[x] = 10
        elif card == 'A':
            cards[x] = 11
        x += 1
    return sum(cards)


# --------------------
# Welcome Message
# --------------------

print("\nHello! Welcome to Blackjack Game!") 

name = input("What is your name?")

print(f"\nHi {name}, are you ready to play?")

ready = input("Are you ready to play? (yes/no)")

ready = ready.lower() 
input_check = None

while input_check != "ok":
    if ready == "yes":
        input_check = "ok"
        print(f"\nPerfect {name}! 🎉 Let's play! Good luck! 😎")
    elif ready == "no":
        input_check = "ok"
        print(f"\nOh!☹️ Ok... See you next time {name}!")
    else:
        ready = input(f"\nSorry {name}, i didn't understand... please try again. Are you ready to play? (yes/no)")
        ready = ready.lower()

time.sleep(3)

while ready == "yes" and try_again == "yes":

    player_choice = "hit"
    dealer_choice = "hit"
    player_cards_total = []
    dealer_cards_total = []
    dealer_score = 0
    first_time = True
    first_cards = True

    # --------------------
    # Reset Deck
    # --------------------

    cards_deck = create_cards_deck()

    while player_choice == "hit" or dealer_choice == "hit":

        if first_cards == True:
            x = 2
        else:
            x = 1

        

        if player_choice == "hit":

            # --------------------
            # Cards distribution and deck cards update
            # --------------------
        
            player_cards = random.sample(cards_deck,x)

            [cards_deck.remove(card) for card in player_cards]

            # --------------------
            # Show the cards to the player
            # --------------------
            player_cards_total = player_cards + player_cards_total

            if not first_cards == True:
                print(f"\n{name}, your new card is:",player_cards)


            time.sleep(3)
            print(f"\n{name}, your cards are:",player_cards_total)

            time.sleep(3)

            player_cards_total_conv = player_cards_total

            player_score = score_calculate(player_cards_total_conv)

            print(f"\n{name}, your score is:",player_score)

            time.sleep(3)

            if player_score <= 21:

                # --------------------
                # Hit or Stand? (Player)
                # --------------------

                print("\nIt's time to choose if you wanna Hit or Stand:")
                player_choice = input("Choose: Hit or Stand?")
                player_choice.lower()

                input_check = ""

                while input_check != "ok":
                    if player_choice == "stand" or player_choice == 'hit' :
                        input_check = "ok"
                        print(f"\nGreat! You chose {player_choice}")
                    else:
                        player_choice = input(f"\nSorry {name}, i didn't understand... please try again. Choose: Hit or Stand?")
                        ready = ready.lower()               

                time.sleep(3)

                if dealer_score < 17:

                # --------------------
                # Cards distribution and deck cards update
                # --------------------

                    dealer_cards = random.sample(cards_deck,x)

                    [cards_deck.remove(card) for card in dealer_cards]

                    dealer_cards_total = dealer_cards + dealer_cards_total

                    dealer_cards_total_conv = dealer_cards_total

                    dealer_score = score_calculate(dealer_cards_total_conv)

                    print("\nDealer's hand:",dealer_cards_total)
                    time.sleep(3)
                    print("\nDealer's score:",dealer_score)
                    time.sleep(3)

                    if dealer_score < 17:
                        print("\nLet's see what was the dealer's choice!")
                        time.sleep(3)
                        print("\nWow! Dealer chose hit!")
                        time.sleep(3)

                    elif dealer_score >= 17 and dealer_score <= 21 :
                        if first_time == True:
                            dealer_choice = "stand"
                            print("\nLet's see what was the dealer's choice!")
                            time.sleep(3)
                            print("\nDealer chose stand!")
                            time.sleep(3)

                            first_time = False
                    else:
                        dealer_choice = "stand"
                        print("Ups! Dealer's hand exploded! 💣💣💣💣💣💣")
                        time.sleep(3)

            else:
                 player_choice = "stand"
                 dealer_choice = "stand"

        elif dealer_choice == "hit":

            if dealer_score < 17:

                dealer_cards = random.sample(cards_deck,x)

                [cards_deck.remove(card) for card in dealer_cards]

                dealer_cards_total = dealer_cards + dealer_cards_total

                dealer_cards_total_conv = dealer_cards_total

                dealer_score = score_calculate(dealer_cards_total_conv)

                print("\nDealer's hand:",dealer_cards_total)
                time.sleep(3)
                print("\nDealer's score:",dealer_score)
                time.sleep(3)

                if dealer_score < 17:
                    print("\nLet's see what was the dealer's choice!")
                    time.sleep(3)
                    print("\nWow! Dealer's have choosed Hit!")
                    time.sleep(3)

                elif dealer_score >= 17 and dealer_score <= 21 :
                    if first_time == True:
                        dealer_choice = "stand"
                        print("\nLet's see what was the dealer's choice!")
                        time.sleep(3)
                        print("\nDealer's have choosen stand!")
                        time.sleep(3)

                        first_time = False
                else:
                    dealer_choice = "stand"
                    print("Ups! Dealer's hand exploded! 💣💣💣💣💣💣")
                    time.sleep(3)

        first_cards = False


    # --------------------
    # Final Score
    # --------------------

    if player_score > 21:
        time.sleep(3)
        print("\nBUST!! 💣💣💣💣💣💣 Your score is higher than 21!")
        time.sleep(3)
        print("\nSorry... 😥 You lose... 😥")
        game_state["player"] += 0
        if dealer_score > 21:
            game_state["dealer"] += 0
    else:
        if player_score > dealer_score:
            time.sleep(3)
            print("\nCongratulations!! 🥇🥇🥇 You win!! ONE POINT TO GRYFFINDOR! 😆")
            game_state["player"] += 1
        elif player_score < dealer_score:
            if dealer_score > 21:
                print("\nCongratulations!! 🥇🥇🥇 You win!! Dealer's hand exploded! 💣💣💣💣💣")
                game_state["player"] += 1
            else:
                time.sleep(3)
                print("\nSorry... 😥 You lose... 😥")
                game_state["dealer"] += 1
        else:
            time.sleep(3)
            print("\nPush!! 😎 It was a draw!")
            time.sleep(3)
            print("\nOne point for each of you!")
            game_state["player"] += 1
            game_state["dealer"] += 1

    time.sleep(3)

    print("\n--------------------")
    print("PONTUATION")
    print(f"\nPlayer Score: {player_score}")
    print(f"Dealer Score: {dealer_score}")
    print("--------------------")
    print(f"Total Player Score: {game_state['player']}")
    print(f"Total Dealer Score: {game_state['dealer']}")
    print("--------------------")

    time.sleep(3)

    print("\nDo you wanna try again?")
    try_again = input("Do you wanna try again?(yes/no)")
    try_again.lower()

    input_check = ""

    while input_check != "ok":
        if try_again == "yes" or try_again == 'no' :
            input_check = "ok"
        else:
            try_again = input(f"\nSorry {name}, i didn't understand... please try again. Do you wanna try again? (yes/no)")
            try_again = try_again.lower()  


if ready == "yes":
    print("\nFINAL SCORE:")

    print("\n--------------------")
    print(f"Player score: {game_state['player']}")
    print(f"Dealer score: {game_state['dealer']}")
    if game_state['player'] > game_state['dealer']:
        print("\nCongratulations!! 🥇🥇🥇 You win!!")
    elif game_state['player'] < game_state['dealer']:
        print("\nGame over! 😡")
    else:
        print("\nPush!! 😎 It was a draw!")
    print("--------------------")
    print("\nSee you next time! ✌️")
```
